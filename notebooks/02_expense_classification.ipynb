# 02: Expense Categorisation via NLP
### Comparing TF-IDF Baselines vs Sentence Transformer Embeddings

This notebook explores transaction categorisation as a multi-class NLP text classification problem. We compare:
1. **TF-IDF + Multinomial Logistic Regression** (High speed baseline)
2. **TF-IDF + Random Forest Classifier** (Non-linear bag-of-words)
3. **Sentence Transformer (`all-MiniLM-L6-v2`) Embeddings + Logistic Classifier** (Semantic dense representations)


```python
import os
import sys
sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.categorisation import train_and_evaluate_all, predict_category
print("Imported categorisation module.")
```

## 1. Train and Benchmark Categorisation Models


```python
results = train_and_evaluate_all('../data/processed/cleaned_transactions.csv')
benchmark_df = results['benchmark']
display(benchmark_df)
```

## 2. Confusion Matrix of the Best Classifier


```python
best_model_key = 'tfidf_lr'
cm = results['models'][best_model_key]['cm']
classes = results['models'][best_model_key]['classes']

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
plt.title(f"Confusion Matrix: {results['models'][best_model_key]['metrics']['model']}", fontsize=13, fontweight='bold')
plt.xlabel("Predicted Category")
plt.ylabel("True Category")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 3. Real-Time Inference on Arbitrary Transaction Strings


```python
test_transactions = [
    "TESCO EXTRA BRAY -€48.72",
    "SPOTIFY AB MONTHLY 10.99 EUR",
    "CIRCLE K MOTORWAY SERVICES CORK",
    "AMZN Mktp UK*93K21 ELECTRONICS",
    "FLYEfit Gym Monthly DD Dublin",
    "Dominos Pizza Galway",
    "ELECTRIC IRELAND DIRECT DEBIT"
]

for tx in test_transactions:
    cat, conf = predict_category(tx, results['models']['tfidf_lr']['pipeline'])
    print(f"{tx:40} -> {cat:28} (Confidence: {conf:.1%})")
```
