## 1. Inspecting Raw Messy Transactions
Real financial exports are messy:
- Mixed date conventions (`%Y-%m-%d`, `%d/%m/%Y`, `%b %d, %Y`, `%Y.%m.%d`)
- Inconsistent casing, trailing whitespace, store numbers, merchant transaction codes
- Mixed amount conventions (negative values, currency symbols `€`, `EUR`, whitespace)
- Missing values and duplicate swipe records


```python
raw_df = pd.read_csv('../data/raw/messy_transactions.csv')
print(f"Total raw rows: {len(raw_df)}")
display(raw_df.head(10))
```

```python
print("Missing values in raw data:")
print(raw_df.isna().sum())
print(f"\nDuplicate rows in raw data: {raw_df.duplicated(subset=['raw_date', 'raw_merchant', 'raw_amount']).sum()}")
```

## 2. Running the Cleaning & Normalization Pipeline


```python
clean_df = clean_transactions(raw_df)
print(f"Cleaned dataset rows: {len(clean_df)}")
display(clean_df.head(10))
```

## 3. Financial Exploratory Visualizations


```python
# Monthly spending distribution
monthly_spend = clean_df.groupby('year_month')['amount'].sum().reset_index()

plt.figure(figsize=(12, 5))
plt.plot(monthly_spend['year_month'], monthly_spend['amount'], marker='o', color='#2563eb', linewidth=2)
plt.xticks(rotation=45)
plt.title("Total Monthly Spending Trajectory (2023 - 2025)", fontsize=14, fontweight='bold')
plt.ylabel("Spending (€)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Category breakdown
plt.figure(figsize=(10, 5))
cat_spend = clean_df.groupby('category')['amount'].sum().sort_values(ascending=False)
sns.barplot(x=cat_spend.values, y=cat_spend.index, palette='Blues_r')
plt.title("Total Spending by Category", fontsize=14, fontweight='bold')
plt.xlabel("Total Spent (€)")
plt.tight_layout()
plt.show()
```
